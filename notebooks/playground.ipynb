```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Example practice data and simple transformations
datasets = [3, 7, 5, 9, 2, 10, 6, 1, 8, 4, 15, 18, 13, 20, 16, 11, 19, 14, 17, 12, 25, 28, 23, 30, 26, 21, 29, 24, 27, 32, 35, 38, 33, 40, 36, 31, 39, 34, 37, 32, 45, 48, 43, 50, 46, 41, 49, 44, 47, 42, 55, 58, 53, 60, 56, 51, 59, 54, 57, 52, 65, 68, 63, 70, 66, 61, 69, 64, 67, 62, 5, 12, 18, 24, 31, 37, 43, 49, 56, 62, 6, 14, 20, 27, 34, 40, 46, 63, 59, 65, 2, 17, 25, 39, 41, 54, 68, 11, 29, 47]
array = np.array(datasets).reshape(10, 10)

print('Dataset shape:', array.shape)
print('First row:', array[0])

filtered = [value for value in datasets if 61 <= value <= 70]
print('Values between 61 and 70:', filtered)
print('Count:', len(filtered))

def calculate_range(data, low, high):
    return [value for value in data if low <= value <= high]

def reshape_dataset(data, rows, cols):
    arr = np.array(data)
    if arr.size != rows * cols:
        raise ValueError('Target shape does not match data length')
    return arr.reshape(rows, cols)

print('Range 10-20:', calculate_range(datasets, 10, 20))
print('Reshaped dataset:')
print(reshape_dataset(datasets[:20], 4, 5))
```
